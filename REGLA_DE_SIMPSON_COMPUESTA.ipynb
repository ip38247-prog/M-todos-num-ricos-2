#REGLA DE SIMPSON COMPUESTA



#####PARA OCUPAR ESTA REGLA TOMAMOS UN $N$ ENTERO PAR,Y SE DIVIDE EL INTERVALO $[a,b]$ EN $N$ SUBINTERVALOS CON $h=\frac{b-a}{2}$


```python
import math
import matplotlib.pyplot as plt
```

```python
def f(x):
    return math.exp(2*x) * math.sin(3*x)
```

```python
a =0
b =2
n = int(input("n: "))
```

```python
h = (b - a) / n
```

```python
x0 = a**2 + b**2
x1 = 0
x2 = 0
```

```python
for i in range(1, n):
    x = a + i * h
    if i % 2 == 0:
        x2 = x2 + f(x)
    else:
        x1 = x1 + f(x)

print(h * (x0 + 2 * x2 + 4 * x1) / 3)
```

```python
cx = []
cy = []
for i in range(201):
    x = a + i * (b - a) / 200
    cx.append(x)
    cy.append(f(x))

plt.plot(cx, cy)
plt.title('Regla de Simpson')
plt.show()
```
