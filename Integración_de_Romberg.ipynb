#Integracion de Romberg

#####Nos siver para calcular el valos de una integral definida $\int_{a}^{b} f(x) \, dx$ cuando no se puede calcular de forma analitica

#####Es mas preciso que la regla del trapesi o la de simpson,converge mas rapido para funciones suaves

#####La funcion $f(x)$ debe ser continua y suave,los limites $a$ y $b$ deben ser finitos y $n$ debe ser un entero positivo

```python
import math

def romberg(f,a,b,n):
    r1=[0]*n
    r2=[0]*n
    h=b-a
    r1[0]=(h/2)*(f(a)+f(b))
    print(f"i = 1: {r1[0]:.10f}")

    for i in range(2,n+1):
        s=0
        m=2**(i-2)
        for k in range(1,m+1):
            s=s+f(a+(k-0.5)*h)
        r2[0]=0.5*(r1[0]+h*s)
        for j in range(2,i+1):
            r2[j-1]=r2[j-2]+(r2[j-2]-r1[j-2])/((4**(j-1))-1)
        fila=""
        for j in range(1,i+1):
            fila=fila+f"{r2[j-1]:.10f} "
        print(f"i={i}:{fila}")
        h = h/2
        for j in range(1,i+1):
            r1[j-1] = r2[j-1]

    return r1[n-1]
```

```python
def f(x):
    return math.sin(x)
res = romberg(f,0,math.pi,6)
print(f"\nResultado:{res:.10f} ")
```
