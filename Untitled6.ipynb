#Cuadratura adaptable

una técnica numérica para calcular integrales definidas $I=\int_a^b f(x)\,dx$
como funciona
1. Aproxima la integral en un subintervalo.
2. Estima el error local.
3. Si el error es mayor que una tolerancia, subdivide ese subintervalo.
4. Repite el proceso recursivamente.

```python
import math
def cuadratura_adaptable(f,a,b,tol,n):
    tolv=[0]*(n+2)
    av=[0]*(n+2)
    hv=[0]*(n+2)
    fa=[0]*(n+2)
    fc=[0]*(n+2)
    fb=[0]*(n+2)
    s=[0]*(n+2)
    l=[0]*(n+2)
    evaluaciones=0
    app=0
    i=1
    tolv[i]=10*tol
    av[i]=a
    hv[i]=(b-a)/2
    fa[i]=f(a)
    fc[i]=f(a+hv[i])
    fb[i]=f(b)
    evaluaciones=evaluaciones+3
    s[i]=hv[i]*(fa[i]+4*fc[i]+fb[i])/3
    l[i]=1
    while i>0:
        fd=f(av[i]+hv[i]/2)
        fe=f(av[i]+3*hv[i]/2)
        evaluaciones=evaluaciones+2
        s1=hv[i]*(fa[i]+4*fd+fc[i])/6
        s2=hv[i]*(fc[i]+4*fe+fb[i])/6
        v1=av[i]
        v2=fa[i]
        v3=fc[i]
        v4=fb[i]
        v5=hv[i]
        v6=tolv[i]
        v7=s[i]
        v8=l[i]
        i=i-1
        if abs(s1+s2-v7)<v6:
            app=app+s1+s2
        else:
            if v8>=n:
                print("nivel excedido")
                return None
            else:
                i=i+1
                av[i]=v1+v5
                fa[i]=v3
                fc[i]=fe
                fb[i]=v4
                hv[i]=v5/2
                tolv[i]=v6/2
                s[i]=s2
                l[i]=v8+1
                i=i+1
                av[i]=v1
                fa[i]=v2
                fc[i]=fd
                fb[i]=v3
                hv[i]=hv[i-1]
                tolv[i]=tolv[i-1]
                s[i]=s1
                l[i]=l[i-1]
    print("solucion:",app)
    print("evaluaciones:",evaluaciones)
    return app
```

usando el algoritmo para $f (x) = (\frac{100}{x^2
})sen(\frac{10}{x})$

```python
def f(x):
    return (100/(x**2))*math.sin(10/x)
cuadratura_adaptable(f,1,3,0.0001,50)
```
