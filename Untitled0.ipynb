```python
import numpy as np
import matplotlib.pyplot as plt
```

grafica de $f(x)=e^{x^2}$ y $p(x)=1+x^2+\frac{x^4}{2!}$

```python
def f(x):
  return np.exp(x**2)
```

```python
f(1)
```

```python
xx=np.linspace(0,1,100)
plt.plot(xx,f(xx))
```

```python
def g(x):
  return 1+x**2+(x**4)/2
```
